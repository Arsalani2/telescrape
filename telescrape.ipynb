## ۰) سکرت‌ها (فقط بار اول)

در ستون چپ، آیکون 🔑 **Secrets** → نام‌های زیر را بسازید، مقدار بدهید و **Notebook access** هرکدام را روشن کنید:

| نام | مقدار |
|---|---|
| TG_API_ID | api id از my.telegram.org |
| TG_API_HASH | api hash از my.telegram.org |
| TG_PHONE | شماره، مثل +98912… (اختیاری) |
| TG_2FA_PASSWORD | رمز دو مرحله‌ای (اگر دارید؛ اختیاری) |
| TS_ENC_KEY | رمز بکاپ سشن (اختیاری) |
| APP_TOKEN | رمز لینک عمومی — فقط برای سلول ۲ب (الفبایی-عدمی، مثلاً ۲۰-۴۰ کاراکتر) |

> ⚠️ سکرت‌ها فقط در پروسه‌ی کرنلِ نوت‌بوک قابل خواندن‌اند؛ سلول لانچر آنها را می‌خواند و به سرور پاس می‌دهد — وارد کردن دستی لازم نیست.
